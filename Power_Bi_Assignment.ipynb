# Introduction to Power BI, Charts, DAX & Creating Reports


Question 1 :
Define Power BI and What are the key components of the Power BI ecosystem? Briefly
explain:
● Power BI Desktop
● Power BI Service
● Power BI Mobile
● Power BI Gateway

Answer :
  - Power BI is a business analytics tool that provides interactive visualizations and business intelligence capabilities with an interface simple enough for end users to create their own reports and dashboards.

  - The key components of the Power BI ecosystem are:

    - Power BI Desktop: This is a free application that we can download and install on our computer. It's where we connect to data sources, transform and model your data, and create reports and visualizations.

    - Power BI Service: This is a cloud-based service (SaaS) where we publish our reports from Power BI Desktop. It allows us to share our reports and dashboards with others, collaborate, and set up refresh schedules. It's the platform for sharing and consuming our insights.

    - Power BI Mobile: This is a set of native mobile apps for Windows, iOS, and Android devices. It lets us view and interact with our Power BI reports and dashboards on the go, so we can stay updated wherever we are.

    - Power BI Gateway: This acts as a bridge, enabling Power BI Service to connect to on-premises data sources. It securely transfers data between your local data and the Power BI cloud service, ensuring your cloud reports have the latest information.

Question 2 :
Compare the following Power BI visuals:
● Pie Chart vs Donut Chart
● Bar Chart vs Column Chart
When would you prefer one over the other? Give one example for each pair.

Answer:
- Pie vs. Donut:
    - Both show proportions of a whole. Use a pie chart for 5 or fewer categories where we want to emphasize individual slices. A donut chart is better if we have slightly more categories or want to display a total value in the center.
    - Example (Pie): Distribution of votes among 4 candidates.
    - Example (Donut): Breakdown of a company's expenses, with total expenditure in the middle.

- Bar vs. Column:
    - Both compare discrete values. Bar charts (horizontal bars) are ideal when category labels are long or we have many categories, as they provide more space. Column charts (vertical bars) are often preferred for showing trends over time or comparing a few categories with short labels.
    - Example (Bar): Comparing the population of several countries with long names.
    - Example (Column): Monthly sales figures for a product.

Question 3 :
Explain the significance of:
● Star schema vs Snowflake schema
● Primary key vs Foreign key in relationships (Power BI)
Why is cardinality important?

Answer:
- Star vs. Snowflake Schema:
    - The star schema is simpler, with a central fact table directly linked to multiple dimension tables. This is great for performance in Power BI because queries are faster with fewer joins.
    - The snowflake schema is more normalized, with dimension tables linked to other dimension tables. This reduces data redundancy but can lead to more complex queries and slower performance due to more joins.

- Primary Key vs. Foreign Key:
    - A primary key uniquely identifies each record in a table.
    - A foreign key is a field in one table that uniquely identifies a row of another table; it's used to link tables.
    - In Power BI, these keys are crucial for establishing relationships between your data tables, allowing us to combine and analyze data from different sources effectively.

- Importance of Cardinality:
Cardinality defines the nature of the relationship between two tables (e.g., one-to-one, one-to-many, many-to-many). It's vital in Power BI because it tells the engine how to correctly aggregate and filter data across related tables. Incorrect cardinality can lead to inaccurate calculations, duplicated data, or performance issues.

Question 4 :
Differentiate between:
● Calculated column vs Measure
Also, define Row context and Filter context with simple examples.

Answer:

- Calculated Column vs. Measure:

    - A calculated column is a physical column added to our data model that computes its value for each row during data refresh, storing the result. A measure, on the other hand, is a dynamic calculation performed on the fly based on the current filter context, and its result is not stored.
    - Example(Calculated Column) - Adding a 'FullName' column by concatenating 'FirstName' and 'LastName' columns.
    - Example(Calculated Measure) - Calculating 'Total Sales' by summing the 'SalesAmount' column.

- Row Context and Filter Context:

    - Row context refers to the evaluation of an expression within the context of a single row in a table. Filter context refers to the set of filters that are applied to the data model, affecting which data is visible and included in calculations.
    - Example(Row Context) - When creating a calculated column like 'Profit = Sales - Cost', the calculation 'Sales - Cost' happens for each individual row using the 'Sales' and 'Cost' values from *that specific row
    - Example(Row Filter) - If I have a measure for 'Total Sales' and I apply a slicer for "Year = 2023", the filter context is now set to only include data from 2023. The 'Total Sales' measure will then sum sales only for that filtered year.

Question 5:
What is the difference between a report and a dashboard in Power BI?

Answer:
  - A report is a multi-page canvas that contains visualizations, text, and interactive elements. It's designed for in-depth exploration and analysis of data.
  - A dashboard is a single-page canvas that provides a high-level overview of key metrics and insights, often pulling visualizations from multiple reports. Dashboards are meant for quick monitoring and "at-a-glance" understanding.
